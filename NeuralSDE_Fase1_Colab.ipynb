## 📐 4. GBM & Ito's Lemma

Model GBM:
$$dS_t = \mu S_t\,dt + \sigma S_t\,dW_t$$

Solusi eksak via **Ito's Lemma** (terapkan $f = \ln S$):
$$S_t = S_0 \cdot \exp\!\Bigl[\bigl(\mu - \tfrac{\sigma^2}{2}\bigr)t + \sigma W_t\Bigr]$$

> **Mengapa ada koreksi $-\sigma^2/2$?**
> Di kalkulus biasa: $(dx)^2 \approx 0$.
> Di kalkulus Itô: $(dW_t)^2 = dt$ *(quadratic variation tidak nol!)*
> Tanpa koreksi ini kita akan **overestimate** nilai ekspektasi harga.

Kalibrasi GBM ke setiap aset, lalu simulasi 1 tahun ke depan.


```python
T      = 1.0
dt     = 1/252
N      = int(T / dt)
N_PATH = 1500

gbm_results = {}

for ticker, d in assets.items():
    ret   = d['returns']
    n_day = 252 if 'JK' in ticker else 365

    mu_hat    = ret.mean() * n_day
    sigma_hat = ret.std()  * np.sqrt(n_day)
    S0        = d['prices'].iloc[-1]

    Z          = np.random.normal(0, 1, (N, N_PATH))
    increments = (mu_hat - 0.5*sigma_hat**2)*dt + sigma_hat*np.sqrt(dt)*Z
    S_paths    = S0 * np.exp(np.vstack([np.zeros(N_PATH),
                                         np.cumsum(increments, axis=0)]))

    S_T_sim  = S_paths[-1].mean()
    S_T_theo = S0 * np.exp(mu_hat * T)

    gbm_results[ticker] = {
        'paths'   : S_paths,
        'mu'      : mu_hat,
        'sigma'   : sigma_hat,
        'S0'      : S0,
        'S_T_sim' : S_T_sim,
        'S_T_theo': S_T_theo,
    }

# Print summary
print(f"{'Ticker':<10} {'μ (ann)':>9} {'σ (ann)':>9} "
      f"{'S0':>10} {'E[S_T] sim':>12} {'E[S_T] theo':>13}")
print("-" * 65)
for t, r in gbm_results.items():
    print(f"{t:<10} {r['mu']:>9.1%} {r['sigma']:>9.1%} "
          f"{r['S0']:>10.2f} {r['S_T_sim']:>12.2f} {r['S_T_theo']:>13.2f}")

print("\n📌 Perhatikan σ crypto (SOL, XRP) vs saham IDX: jauh lebih besar!")
print("   Volatility tinggi = lebih susah dimodel dengan GBM sederhana")
```

## 🔁 6. Euler–Maruyama

Ketika tidak ada solusi eksak, kita gunakan diskretisasi **Euler–Maruyama**:

$$S_{t+\Delta t} = S_t + \underbrace{\mu S_t\,\Delta t}_{\text{drift}} + \underbrace{\sigma S_t\,\Delta W_t}_{\text{diffusion}}$$

> 🔑 **Koneksi ke Neural SDE:**
> Di Fase 2 nanti, kita ganti $\mu$ dan $\sigma$ dengan **neural network** $f_\theta$ dan $g_\theta$.
> `torchsde` melakukan loop Euler–Maruyama ini tapi dengan **automatic differentiation**
> sehingga gradien bisa mengalir balik melalui SDE solver.


```python
# Demo Euler-Maruyama vs Exact untuk SEMUA ASET
fig, axes = plt.subplots(len(assets), 2, figsize=(14, 4.5 * len(assets)))
fig.suptitle('1 Path: Exact vs E-M & Terminal Distribution S_T untuk Semua Aset', fontsize=16, fontweight='bold')

for i, ticker in enumerate(assets.keys()):
    res  = gbm_results[ticker]
    col  = assets[ticker]['color']
    mu   = res['mu']
    sig  = res['sigma']
    S0   = res['S0']

    S_em    = np.zeros((N + 1, N_PATH))
    S_em[0] = S0
    for t in range(N):
        dW         = np.random.normal(0, np.sqrt(dt), N_PATH)
        S_em[t+1]  = S_em[t] + mu*S_em[t]*dt + sig*S_em[t]*dW

    print(f"Demo: {assets[ticker]['label']}")
    print(f"{'Method':<20} {'E[S_T]':>10} {'vs Teori':>10}")
    print("-" * 42)
    print(f"{'GBM Exact':<20} {res['S_T_sim']:>10.2f} "
          f"{(res['S_T_sim']-res['S_T_theo'])/res['S_T_theo']:>+9.3%}")
    print(f"{'Euler-Maruyama':<20} {S_em[-1].mean():>10.2f} "
          f"{(S_em[-1].mean()-res['S_T_theo'])/res['S_T_theo']:>+9.3%}")
    print(f"{'Teori':<20} {res['S_T_theo']:>10.2f}\n")

    # Plot single path comparison
    axes[i, 0].plot(time_ax, res['paths'][:, 0],
                 label='GBM Exact', color=col, linewidth=1.5)
    axes[i, 0].plot(time_ax, S_em[:, 0],
                 label='Euler-Maruyama', color='coral', linewidth=1.5, linestyle='--')
    axes[i, 0].set_title(f'{assets[ticker]["label"].split(" ", 1)[1]} - 1 Path: Exact vs E-M')
    axes[i, 0].set_xlabel('Tahun')
    axes[i, 0].legend()

    # Plot histogram
    axes[i, 1].hist(res['paths'][-1], bins=60, density=True,
                 alpha=0.6, color=col, label='GBM Exact S_T')
    axes[i, 1].hist(S_em[-1], bins=60, density=True,
                 alpha=0.6, color='coral', label='E-M S_T')
    axes[i, 1].set_title(f'{assets[ticker]["label"].split(" ", 1)[1]} - Terminal Dist S_T')
    axes[i, 1].set_xlabel('S_T')
    axes[i, 1].legend()

plt.tight_layout(rect=[0, 0, 1, 0.98])
plt.savefig('phase1_euler_maruyama_all.png', dpi=130, bbox_inches='tight')
plt.show()
```

## 🔬 7. Stylized Facts — Perbandingan Semua Aset vs GBM

Tiga "stylized facts" yang tidak bisa ditangkap GBM:

| Fact | Definisi | Implikasi |
|------|----------|-----------|
| **Fat tails** | Kurtosis >> 3 | Crash lebih sering dari prediksi GBM |
| **Vol clustering** | ACF(\|r\|²) > 0 | Periode tenang & chaos bergantian |
| **Leverage effect** | Skewness < 0 | Turun lebih ekstrem dari naik |


```python
def compute_stylized_facts(returns, n_day=252):
    """Hitung semua stylized facts dari array daily log-returns."""
    ann_vol  = returns.std() * np.sqrt(n_day)
    ann_ret  = returns.mean() * n_day
    kurt     = stats.kurtosis(returns)        # excess kurtosis (Normal = 0)
    skew     = stats.skew(returns)
    # Volatility clustering: ACF of squared returns at lag 1
    r2 = returns**2
    acf_vol = np.corrcoef(r2[:-1], r2[1:])[0, 1]
    # Leverage effect: correlation between r_t and |r_{t+1}|
    lev = np.corrcoef(returns[:-1], np.abs(returns[1:]))[0, 1]
    # VaR 1% historical
    var1 = np.percentile(returns, 1)
    return dict(ann_ret=ann_ret, ann_vol=ann_vol,
                kurt=kurt, skew=skew,
                acf_vol=acf_vol, lev=lev, var1=var1)

# Tabel perbandingan
rows = []
for ticker, d in assets.items():
    n_day = 252 if 'JK' in ticker else 365
    sf = compute_stylized_facts(d['returns'], n_day)
    rows.append({'Ticker': ticker, 'Nama': d['label'].split(' ', 1)[1], **sf})

df_sf = pd.DataFrame(rows).set_index('Ticker')
df_sf.columns = ['Nama', 'Ann.Return', 'Ann.Vol',
                  'Kurtosis', 'Skewness', 'ACF vol²',
                  'Leverage', 'VaR 1%']

# Format display
fmt = {'Ann.Return': '{:.1%}', 'Ann.Vol': '{:.1%}',
       'Kurtosis': '{:.2f}',   'Skewness': '{:.3f}',
       'ACF vol²': '{:.3f}',   'Leverage': '{:.3f}',
       'VaR 1%': '{:.3%}'}
display(df_sf.style.format(fmt).background_gradient(
    subset=['Kurtosis','ACF vol²'], cmap='YlOrRd'))

print("\n📌 Interpretasi kolom:")
print("  Kurtosis  > 0  → fat tails (semakin besar = semakin 'berbahaya')")
print("  Skewness  < 0  → leverage effect (crash > rally)")
print("  ACF vol²  > 0  → volatility clustering")
print("  VaR 1%         → worst daily loss pada bottom 1%")
```

## ⚖️ 10. P Measure vs Q Measure

Konsep **wajib** sebelum Fase 2 — Neural SDE beroperasi di salah satu dari dua "dunia" ini.

| | **P Measure** (Real-World) | **Q Measure** (Risk-Neutral) |
|---|---|---|
| Nama lain | Physical / historical | Risk-neutral / pricing |
| Drift | μ = return historis | r = risk-free rate |
| Dipakai untuk | Forecasting, VaR, scenario gen | Pricing opsi & derivatif |
| GBM | dS = **μ** S dt + σ S dW | dS = **r** S dt + σ S dW̃ |

> **Untuk project ini:** kita di **P measure** — tujuannya scenario generation & VaR.
> Keduanya dihubungkan oleh **Girsanov's theorem**: ganti measure = ganti drift.


```python
# ── Demonstrasi numerik: P vs Q measure ──────────────────────────────────────
N_PATH = 2000
np.random.seed(42)

def gbm_paths(S0, drift, sigma, dt, Z, paths):
    log_inc = (drift - 0.5*sigma**2)*dt + sigma*np.sqrt(dt)*Z
    S = S0 * np.exp(np.cumsum(log_inc, axis=0))
    return np.vstack([np.full(paths, S0), S])

for ticker in assets.keys():
    # 1. Parameter spesifik untuk Saham vs Kripto
    trading_days = 252 if 'JK' in ticker else 365
    
    ret  = assets[ticker]['returns']
    S0   = assets[ticker]['prices'].iloc[-1]

    mu_P  = ret.mean() * trading_days           # drift historis (P measure)
    sigma = ret.std()  * np.sqrt(trading_days)  # volatilitas sama di keduanya
    r_Q   = 0.0625                              # BI Rate 2026 approx (Q measure)

    T, dt = 1.0, 1/trading_days
    N     = int(T / dt)

    # Generate Noise (Z) beda panjang untuk Saham (N=252) dan Crypto (N=365)
    Z = np.random.normal(0, 1, (N, N_PATH))

    # 2. Simulasi
    S_P = gbm_paths(S0, mu_P, sigma, dt, Z, N_PATH)
    S_Q = gbm_paths(S0, r_Q,  sigma, dt, Z, N_PATH)

    time_ax = np.linspace(0, T, N + 1)
    
    # 3. Visualisasi
    fig, axes = plt.subplots(1, 2, figsize=(13, 5))
    fig.suptitle(f'P vs Q Measure — {assets[ticker]["label"]}',
                 fontsize=13, fontweight='bold')

    for ax, S, label, col, drift in [
        (axes[0], S_P, 'P Measure (Historical drift)', 'steelblue', mu_P),
        (axes[1], S_Q, 'Q Measure (Risk-free rate)',   'coral',     r_Q),
    ]:
        ax.plot(time_ax, S[:, :300], alpha=0.06, color=col, linewidth=0.7)
        ax.plot(time_ax, S.mean(axis=1), color='black', linewidth=2.5,
                label=f'E[S_T] = {S[-1].mean():.0f}', zorder=5)
        ax.fill_between(time_ax,
                        np.percentile(S, 5,  axis=1),
                        np.percentile(S, 95, axis=1),
                        alpha=0.15, color=col)
        ax.set_title(f'{label}\ndrift = {drift:.1%}/tahun')
        ax.set_xlabel('Tahun')
        ax.set_ylabel('Harga')
        ax.legend(fontsize=9)

    plt.tight_layout()
    plt.show()

    # 4. Ringkasan numerik untuk setiap aset
    print(f"{'':30} {'P Measure':>12} {'Q Measure':>12}")
    print("-" * 56)
    print(f"{'Drift':<30} {mu_P:>12.2%} {r_Q:>12.2%}")
    print(f"{'Volatility':<30} {sigma:>12.2%} {sigma:>12.2%}")
    print(f"{'E[S_T]':<30} {S_P[-1].mean():>12.0f} {S_Q[-1].mean():>12.0f}")
    print(f"{'5th pctile S_T':<30} {np.percentile(S_P[-1],5):>12.0f} {np.percentile(S_Q[-1],5):>12.0f}")
    print(f"{'95th pctile S_T':<30} {np.percentile(S_P[-1],95):>12.0f} {np.percentile(S_Q[-1],95):>12.0f}")
    print("=" * 60 + "\n")
```

1. Dari 7 aset yang kamu analisis, mana yang paling susah dimodel dengan GBM dan kenapa? (lihat kolom kurtosis + ACF vol²) 
    Aset yang paling susah dimodel dengan GBM kemungkinan besar XRP, karena kombinasi kurtosis sangat tinggi (~16.91) dan volatility clustering yang kuat berarti distribusi return-nya jauh dari asumsi Gaussian dengan volatilitas konstan yang dipakai GBM. GBM mengasumsikan increments independen dan variance konstan, sedangkan XRP menunjukkan heavy tails dan regime volatility yang berubah-ubah. 
    Secara intuitif, market crypto lebih sering mengalami jump besar, panic movement, dan speculative bursts dibanding saham bank besar.

2. BRIS punya kurtosis 12.69 — jauh lebih tinggi dari BMRI (4.29) dan BBRI (5.65). Menurutmu kenapa? Ada yang spesial dari saham itu secara fundamental?

Jadi secara fundamental, BRIS bukan sekadar bank syariah, tetapi juga saham dengan struktur market yang lebih mudah menghasilkan tail events.
